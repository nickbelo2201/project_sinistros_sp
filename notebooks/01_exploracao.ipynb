# Importando os dados da tabela

## função de leitura dos dados

```python
import pandas as pd
import matplotlib.pyplot as plt

# função de leitura csv
def ler_tabela(nome):
    a = pd.read_csv(f"../data/raw/{nome}_2022-2024.csv", 
                    sep=";", encoding="latin-1", decimal=",", low_memory=False)
    b = pd.read_csv(f"../data/raw/{nome}_2025-2026.csv"
                    , sep=";", encoding="latin-1", decimal=",", low_memory=False)
    assert set(a.columns) == set(b.columns), f"Colunas diferentes em {nome}"
    df = pd.concat([a, b], ignore_index=True)
    assert len(df) == len(a) + len(b) 
    print(nome, a.shape, b.shape, "->", df.shape)
    return df

pessoas = ler_tabela("pessoas")
sinistros = ler_tabela("sinistros")
veiculos = ler_tabela("veiculos")
```

## Armazenando tabelas em parquet

```python
tabelas = {
    "pessoas": pessoas,
    "sinistros": sinistros,
    "veiculos": veiculos,
}

for nome, df in tabelas.items():
    df.to_parquet(f"../data/processed/{nome}.parquet", index=False)
    print(nome, df.shape)
```
