# Importando os dados da tabela

```python
import pandas as pd
import matplotlib.pyplot as plt

def ler(caminho):
    return pd.read_csv(
        caminho, sep=";", encoding="latin-1", decimal=",", low_memory=False)

pessoas_a = ler("../data/raw/pessoas_2022-2024.csv")
pessoas_b = ler("../data/raw/pessoas_2025-2026.csv")

print(pessoas_a.shape, pessoas_b.shape)
print(set(pessoas_a.columns) == set(pessoas_b.columns))

pessoas = pd.concat([pessoas_a, pessoas_b], ignore_index=True)
print(pessoas.shape)
```
